```python
import matplotlib.pyplot as plt
import numpy as np
```

## Basic line graph with numpy and matplotlib

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 25, 30])

plt.plot(x, y)
plt.title("Basic line plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

## Basic line graph with numpy and matplotlib

```python
from matplotlib.lines import lineStyles
x = np.array([1, 2, 3, 4])
y = np.array([3, 8, 1, 10])
plt.plot(y, linestyle="dotted", color="green", marker="o")
plt.title("Styled Line Example")
plt.show()
```

## Mulitpe lines in one graph with numpy and matplotlib

```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([1, 3, 5, 7])
plt.plot(x, y1, label="line 1", color="red", marker="s")
plt.plot(x, y2, label="line 2",  linestyle="--", color="blue", marker=".")
plt.title("Multiple Lines Example")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.legend()
plt.show()
```

## Bar Chart with Numpy and Matplot

```python
x = np.array(["Salary", "House Bills", "Fare", "Food"])
y = np.array([15, 9, 1, 7])

plt.bar(x, y, color="orange")
plt.title("Bar Char Example")
plt.xlabel("Categories")
plt.xlabel("Values")
plt.show()
```

## Scatter Plot with Numpy and Matplot

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 1, 8, 3])

plt.scatter(x, y, color="purple", s=100)
plt.title("Scatter Plot Example")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

## Histogram  with Numpy and Matplot

```python
data = np.array([3, 4, 5, 6, 8, 8, 9, 12])

plt.hist(data, color="skyblue", edgecolor="black")
plt.title("Histogram  Example")
plt.xlabel("X Values")
plt.ylabel("Y Count")
plt.show()
```

# Subplots(Multiple Graphs Together)


```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([3, 6, 9, 12])

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.plot(x, y1, color="red")
plt.title("Plot 1")


plt.subplot(1, 2, 2)
plt.bar(x, y2, color="blue")
plt.title("Plot 2")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-np.pi/2, np.pi/2, 31)
y = np.cos(x)**3

# 1) remove points where y > 0.7
x2 = x[y <= 0.7]
y2 = y[y <= 0.7]

# 2) mask points where y > 0.7
y3 = np.ma.masked_where(y > 0.7, y)

# 3) set to NaN where y > 0.7
y4 = y.copy()
y4[y3 > 0.7] = np.nan

plt.plot(x*0.1, y, 'o-', color='lightgrey', label='No mask')
plt.plot(x2*0.4, y2, 'o-', label='Points removed')
plt.plot(x*0.7, y3, 'o-', label='Masked values')
plt.plot(x*1.0, y4, 'o-', label='NaN values')
plt.legend()
plt.title('Masked and NaN data')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib.ticker import LinearLocator

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

# Make data.
X = np.arange(-5, 5, 0.25)
Y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(X, Y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

# Plot the surface.
surf = ax.plot_surface(X, Y, Z, cmap="coolwarm",
                       linewidth=0, antialiased=False)

# Customize the z axis.
ax.set_zlim(-1.01, 1.01)
ax.zaxis.set_major_locator(LinearLocator(10))
# A StrMethodFormatter is used automatically
ax.zaxis.set_major_formatter('{x:.02f}')

# Add a color bar which maps values to colors.
fig.colorbar(surf, shrink=0.5, aspect=5)

plt.show()
```
