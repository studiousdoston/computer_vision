```python
import cv2
```

```python
from google.colab.patches import cv2_imshow
image = cv2.imread("/content/cat.png")
cv2_imshow(image)
```

```python
print("Shape:", image.shape)
# 1200 => height
# 1200 => width
# 3    => color channels (Red, Green, Blue) |BGR for OpenCV|
print("Height:", image.shape[0])
print("Width:", image.shape[1])
print("Color channels:", image.shape[2])
```

```python
(b, g, r) =image[100, 50]
print("Blue:", b, "Green:", g, "Red:", r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized  = cv2.resize(gray, (300, 300))
cv2_imshow(resized)
```

```python
cropped = gray[120:500, 200:620]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(cropped, cv2.ROTATE_90_COUNTERCLOCKWISE)
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(cropped , 1)
cv2_imshow(flipped)
```

```python
cv2.imwrite("flipped_cat.png", flipped)
```

# OpenCV video streaming codes

```python
import cv2

cap = cv2.VideoCapture(0)

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")

while True:
  success, frame = cap.read()

  if not success:
    break

  gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

  faces = face_cascade.detectMultiScale(gray, 1.5, 4)

  for(x, y, w, h) in faces:
    cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

  cv2.imshow("Real-time Face Detection", frame)

  if cv2.waitKey(1) & 0xFF == ord('q'):
    break

cap.release()
cv2.destroyAllWindows()

# ------------------------------------------------------------

# url = 'rtsp://192.168.1.164:8001/video' # stream_path
''''
cap = cv2.VideoCapture(0)
print(cap)

while True:
  success, frame = cap.read() # True, False

  if not success:
    break

  gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
  blurred = cv2.GaussianBlur(frame, (15, 15), 0)
  edges1 = cv2.Canny(frame, 100, 200)
  edges2 = cv2.Canny(frame, 50, 100)

  flipped = cv2.flip(frame, -1)

  # cv2.imshow("Original", frame)
  # cv2.imshow("Gray", gray)
  # cv2.imshow("blurred:", blurred) # Noise
  # cv2.imshow("edges1:", edges1)
  # cv2.imshow("edges2:", edges2)

  cv2.imshow("flipped:", flipped)

  if cv2.waitKey(1) &  0xFF == ord('q'):
    break

cap.release()
cv2.destroyAllWindows()


'''
```
